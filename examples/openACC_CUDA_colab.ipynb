### Hello World

```python
%cd /content/OpenACC-CUDA-beginners/examples/OpenACC/hello-world/
```

```python
!ls
```

```python
#include <stdio.h>
#include <stdlib.h>
#include <omp.h>

#define N 1000000

int main() {
    float *A = (float*)malloc(N * sizeof(float));
    float *B = (float*)malloc(N * sizeof(float));
    float *C = (float*)malloc(N * sizeof(float));

    // Initialize vectors
    for (int i = 0; i < N; i++) {
        A[i] = i * 1.0f;
        B[i] = (N - i) * 1.0f;
    }

    double start = omp_get_wtime();

    // OpenACC parallel region with copy-in and copy-out clauses
    #pragma acc kernels loop copyin(A[0:N], B[0:N]) copyout(C[0:N])
    for (int i = 0; i < N; i++) {
        C[i] = A[i] + B[i];
    }

    double end = omp_get_wtime();

    // Verify results
    int correct = 1;
    for (int i = 0; i < N; i++) {
        if (C[i] != N * 1.0f) {
            correct = 0;
            break;
        }
    }

    if (correct)
        printf("OpenACC Vector Addition Successful! Time: %f seconds\n", end - start);
    else
        printf("Computation error!\n");

    free(A);
    free(B);
    free(C);

    return 0;
}
```
